```python
import re

def matrix_chain_order(dimensions):
    n = len(dimensions)
    p = [dimensions[0][0]]

    for rows, cols in dimensions:
        p.append(cols)

    m = [[0] * n for _ in range(n)]

    for chain_len in range(2, n + 1):
        for i in range(n - chain_len + 1):
            j = i + chain_len - 1
            m[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + p[i] * p[k + 1] * p[j + 1]
                )

                if cost < m[i][j]:
                    m[i][j] = cost

    return m[0][n - 1]

def read_dimensions(filename):
    with open(filename, "r") as f:
        text = f.read()

    numbers = list(map(int, re.findall(r"\d+", text)))

    dimensions = []
    for i in range(0, len(numbers), 2):
        dimensions.append((numbers[i], numbers[i + 1]))

    return dimensions

dims = read_dimensions("10 (2).txt")
result = matrix_chain_order(dims)

print("Minimum multiplications:", result)
```
